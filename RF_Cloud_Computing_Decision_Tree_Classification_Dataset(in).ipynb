#### Import Libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

#### Importing the data


```python
df = pd.read_csv('/content/Cloud_Computing_Decision_Tree_Classification_Dataset(in).csv')
```

##### Define categorical and numerical features

```python
categorical_features = ["Cloud_Service","Traffic_Level","Instance_ID"]
Numerical_features = ["CPU_Utilization","Memory_Utilization_%","Network_Bandwidth_Mbps","Response_Time_ms","Active_Users","Request_Rate_per_sec","Energy_Consumption_kWh","VM_Count"]
```

```python
X = df.drop(["Resource_Scaling_Decision"],axis = 1)
Y = df["Resource_Scaling_Decision"]
```

#### Converting text into numerical through onehotencoder

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown = "ignore"),
         categorical_features)],
         remainder='passthrough'

)
```

```python
X_processed = preprocessor.fit_transform(X)
```

#### Splitting the data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X_processed,Y,test_size=0.2,random_state=42,stratify =Y)
```

#### Prepare the model

```python
classifier = RandomForestClassifier(n_estimators=100, criterion = 'gini',random_state=42)
```

#### Train the model

```python
classifier.fit(X_train,Y_train)
```

#### Test the model

```python
Y_pred = classifier.predict(X_test)
```

```python
print("accuracy",accuracy_score(Y_test,Y_pred))
```

```python
print("confusion_matrix")
print(confusion_matrix(Y_test,Y_pred))
```

```python
print("classification_report")
print(classification_report(Y_test,Y_pred))
```
