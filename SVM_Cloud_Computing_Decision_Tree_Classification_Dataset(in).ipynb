#### Import the Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, accuracy_score,classification_report
```

#### Importing the dataset

```python
df=pd.read_csv('/content/Cloud_Computing_Decision_Tree_Classification_Dataset(in).csv')
```

#### Define x and y

```python
x=df.drop('Resource_Scaling_Decision',axis=1)
### x=df.drop('Instance_ID',axis=1)
y=df['Resource_Scaling_Decision']
```

```python
y.value_counts()
```

#### Define categorical_features and numerical_features

```python
categorical_features = ["Cloud_Service","Traffic_Level","Instance_ID"]
Numerical_features = ["CPU_Utilization","Memory_Utilization_%","Network_Bandwidth_Mbps","Response_Time_ms","Active_Users","Request_Rate_per_sec","Energy_Consumption_kWh","VM_Count"]
```

#### Converting text -----> numeric

```python
preprocessor=ColumnTransformer(
    transformers=[
        ("cat",OneHotEncoder(handle_unknown="ignore"),
         categorical_features)],
        remainder="passthrough"

    )
```

```python
x_processed=preprocessor.fit_transform(x)
```

#### Train_test_split

```python
x_train,x_test,y_train,y_test=train_test_split(x_processed,y,test_size=0.2,random_state=42)
```

#### Define the model and train it

```python
classifier =SVC(kernel='sigmoid',random_state=0)
classifier.fit(x_train,y_train)
```

#### test the model

```python
y_pred=classifier.predict(x_test)
```

```python
print("confusion_matrix")
cm=confusion_matrix(y_pred,y_test)
print(cm)
```

```python
print("classification_report")
classification_report(y_pred,y_test)
```

```python
print("Accuracy_score")
accuracy_score(y_pred,y_test)
```
