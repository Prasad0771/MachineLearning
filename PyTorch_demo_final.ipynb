# Storage

```python
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
points.storage()
```

```python
# index into a storage manually
points_storage = points.storage()
points_storage[0]
```

```python
points.storage()[4]
```

```python
# changing the value of a storage leads to changing the content of its referring tensor
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
points
```

```python
points_storage = points.storage()
points_storage[0] = 2.0
points
```

```python
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
points.storage()
```

```python
points.storage_offset()
```

```python
points.stride()
```

```python
second_point.storage_offset()
```

```python
second_point.size()
```

```python
second_point.stride()
```

```python
# changing the subtensor will have a side effect on the original tensor
second_point[0] = 10.0
points
```

```python
# clone the subtensor into a new tensor
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
second_point = points[1].clone()
second_point[0] = 10.0
points
```

```python
points_t = points.t()
points_t
```

```python
# verify that the two tensors share the same storage
id(points.storage()) == id(points_t.storage())
```

```python
points.stride()
```

# Working with images

```python
import imageio
img_arr = imageio.imread('bobby.jpg')
img_arr.shape
```

```python
# use the tensor’s permute method to get to an appropriate layout.
# Given an input tensor H × W × C
# pytorch requirement C X H X W
img = torch.from_numpy(img_arr)
out = img.permute(2, 0, 1)
```

```python
# we can preallocate a tensor of appropriate size and fill it with images loaded from a directory
batch_size = 3
batch = torch.zeros(batch_size, 3, 256, 256, dtype=torch.uint8)
```

```python
filenames = ['cat1.png', 'cat2.png', 'cat3.png']
for i, filename in enumerate(filenames):
    img_arr = imageio.imread(filename)
    img_t = torch.from_numpy(img_arr)
    img_t = img_t.permute(2, 0, 1)
    img_t = img_t[:3] # <1>
    batch[i] = img_t
```

```python
# cast a tensor to floating-point and normalize the values of the pixels
batch = batch.float()
batch /= 255.0
```
